```python
from pathlib import Path

ds = VideoFolderDataset("C:/Users/caovy/Desktop/UCF", "v3")


stats = {
    "total_samples": 0,
    "class_count": defaultdict(int),
    "fps_count": defaultdict(int),
    "resolution": defaultdict(int),
    "duration": defaultdict(int)
}

for i, (stream_info, target) in tqdm(enumerate(ds), total=len(ds), colour="red"):
    try:
        # Lấy đường dẫn file video
        path = ds.samples[i][0]

        # Lấy tên thư mục chứa file video (tức là class)
        video_class = Path(path).parent.name  

        # Lấy thông tin video
        fps: None | int = getattr(stream_info, "frame_rate", None)
        num_frames: None | int = getattr(stream_info, "num_frames", None)
        resolution: Tuple[int, int] = (
            getattr(stream_info, "width", 0), 
            getattr(stream_info, "height", 0)
        )

        # Tính thời lượng video nếu có đủ dữ liệu
        length: None | int = (num_frames // fps) if (fps and num_frames and fps > 0) else None

        # Tạo dictionary metadata
        metadata: Dict[str, Any] = VideoMetadata(fps, length, resolution).to_dict()

        # Cập nhật stats
        stats["total_samples"] += 1
        stats["class_count"][video_class] += 1
        if fps:
            stats["fps_count"][fps] += 1
        if resolution:
            stats["resolution"][resolution] += 1
        if length:
            stats["duration"][length] += 1

        # In ra theo format yêu cầu


    except Exception as e:
        print(f"[ERROR] Không thể xử lý video: {path} - {e}")
print(f"{{{metadata}, {video_class}}}")
```
