# Import libraries

```python
from typing import Tuple, Dict, Any
from collections import defaultdict

from tqdm import tqdm

from AI.src.data.model import VideoMetadata
from AI.src.data.dataset import datasets
from AI.src.utils.visualize_dataset import prompt_dataset_statistics, plot_dataset_statistics
```

# UCF crime

```python
ds = datasets["VideoFolderDataset"]("/home/trong/Downloads/test", "v3")

stats = {
    "total_samples": 0,
    "class_count": defaultdict(int),
    "fps_count": defaultdict(int),
    "resolution": defaultdict(int),
    "duration": defaultdict(int)
}

for i, (stream_info, target) in tqdm(enumerate(ds), total=len(ds), colour="red"):
    class_name = ds.classes[target]
    fps: None | int = getattr(stream_info, "frame_rate", None)
    num_frames: None | int = getattr(stream_info, "num_frames", None)
    resolution: Tuple[int, int] = (getattr(stream_info, "width", 0), getattr(stream_info, "height", 0))

    if not (fps is None and num_frames is None) and fps > 0:
        length: int = num_frames // fps
    else:
        length: None = None

    metadata: Dict[str, Any] = VideoMetadata(fps, length, resolution).to_dict()
    
    # Update stats
    stats["total_samples"] += 1
    stats["class_count"][class_name] += 1
    
    if fps is not None:
        stats["fps_count"][fps] += 1
    
    if resolution != (0, 0):
        stats["resolution"][resolution] += 1
    
    if length is not None:
        stats["duration"][length] += 1

# from pprint import pprint as pp
# pp(stats)
print(prompt_dataset_statistics(stats))
plot_dataset_statistics(ds, fpath="../fig/Dataset/UCF/")
```
